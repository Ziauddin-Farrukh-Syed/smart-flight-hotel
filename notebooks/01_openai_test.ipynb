```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)

from backend.models.travel import TravelRequest
from backend.agents.travel_agent import plan_trip
```

```python
request = TravelRequest(
    origin="London",
    destination="New York",
    departure_date="2026-10-02",
    return_date="2026-10-10",
    passengers=1,
    cabin_class="business",
    max_budget=2500,
    max_stops=1,
    hotel_min_rating=4,
)

options = plan_trip(request)

print(f"Found {len(options)} options")

for option in options:
    print(
        option["flight"].airline,
        option["flight"].flight_number,
        " + ",
        option["hotel"].name,
        " = £",
        option["total_cost"],
    )
```

```python
for index, option in enumerate(options, start=1):

    flight = option["flight"]
    hotel = option["hotel"]

    print(f"\nOPTION {index}")
    print("-" * 40)

    print(
        f"Flight: {flight.airline} "
        f"{flight.flight_number}"
    )

    print(
        f"Route: {flight.origin} → "
        f"{flight.destination}"
    )

    print(f"Cabin: {flight.cabin_class}")
    print(f"Stops: {flight.stops}")
    print(f"Flight: £{option['flight_cost']:.2f}")

    print(f"Hotel: {hotel.name}")
    print(f"Hotel rating: {hotel.rating}")
    print(f"Nights: {option['nights']}")
    print(f"Hotel: £{option['hotel_cost']:.2f}")

    print(f"TOTAL: £{option['total_cost']:.2f}")
    print(
        f"Remaining budget: "
        f"£{option['remaining_budget']:.2f}"
    )
```
