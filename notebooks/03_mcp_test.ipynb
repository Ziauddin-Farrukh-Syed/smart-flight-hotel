```python
from mcp import ClientSession
from mcp.client.streamable_http import streamable_http_client
```

```python
async with streamable_http_client(
    "http://127.0.0.1:8000/mcp"
) as (
    read_stream,
    write_stream,
):

    async with ClientSession(
        read_stream,
        write_stream,
    ) as session:

        await session.initialize()

        print("MCP connection successful!")
```

```python
async with streamable_http_client(
    "http://127.0.0.1:8000/mcp"
) as (
    read_stream,
    write_stream,
):

    async with ClientSession(
        read_stream,
        write_stream,
    ) as session:

        await session.initialize()

        tools_result = await session.list_tools()

        print(
            "Number of tools:",
            len(tools_result.tools)
        )

        for tool in tools_result.tools:
            print()
            print("NAME:", tool.name)
            print("DESCRIPTION:", tool.description)
```

```python
async with streamable_http_client(
    "http://127.0.0.1:8000/mcp"
) as (
    read_stream,
    write_stream,
):

    async with ClientSession(
        read_stream,
        write_stream,
    ) as session:

        await session.initialize()

        result = await session.call_tool(
            "search_flights_mcp",
            {
                "origin": "London",
                "destination": "New York",
                "departure_date": "2026-10-02",
                "cabin_class": "business",
            },
        )

        print(result)
```

```python
async with streamable_http_client(
    "http://127.0.0.1:8000/mcp"
) as (
    read_stream,
    write_stream,
):

    async with ClientSession(
        read_stream,
        write_stream,
    ) as session:

        await session.initialize()

        tools_result = await session.list_tools()

        print(
            "Number of MCP tools:",
            len(tools_result.tools)
        )

        for tool in tools_result.tools:
            print("-", tool.name)

        flight_result = await session.call_tool(
            "search_flights_mcp",
            {
                "origin": "London",
                "destination": "New York",
                "departure_date": "2026-10-02",
                "cabin_class": "business",
            },
        )

        print()
        print("FLIGHT RESULT")
        print(flight_result.content)

        hotel_result = await session.call_tool(
            "search_hotels_mcp",
            {
                "destination": "New York",
                "hotel_min_rating": 4.0,
            },
        )

        print()
        print("HOTEL RESULT")
        print(hotel_result.content)

        print()
        print("PHASE 6 MCP TEST PASSED")
```
